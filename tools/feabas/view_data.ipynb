```python
import tensorstore as ts
import matplotlib.pyplot as plt
```

```python
dataset_path = "/home/ubuntu/SmartEM-Processing-Workflows/example_working_dir/meirovitch2025/aligned_tensorstore"

ds = ts.open({
    "driver": "neuroglancer_precomputed",
    "kvstore": {"driver": "file", "path": dataset_path}
}).result()
```

```python
img = ds[:, :, 0, 0].read().result()  # z=0, channel=0

plt.imshow(img, cmap="gray")
plt.axis("off")
plt.show()
```
