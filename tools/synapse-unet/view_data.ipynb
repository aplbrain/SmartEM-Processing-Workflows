# Visualize a volume of inference

```python
import numpy as np
import h5py as h5
import matplotlib.pyplot as plt
```

Note before starting:

The inference files may not open if you ran the docker container as root. You may need to chown these files so that they belong to your username.

```python
seg_vol = np.array(h5.File("/home/ubuntu/code/SmartEM-Processing-Workflows/example_working_dir/meirovitch2025/synapse_unet_outputs/result_txy.h5", mode='r')['vol0'])
image_vol = np.array(h5.File("/home/ubuntu/code/SmartEM-Processing-Workflows/example_working_dir/meirovitch2025/image_volume.h5", mode='r')['raw'])
```

```python
seg = np.squeeze(seg_vol)
seg.shape
```

```python
frame = 50
plt.figure(1, figsize=(12,12))
plt.imshow(image_vol[frame,:,:], cmap='gray')
plt.imshow(seg[frame,:,:], cmap='tab20', alpha=0.2)
```
